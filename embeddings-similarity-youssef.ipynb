# Embeddings and Semantic Similarity

## Objective
Understand how text is converted into numerical vectors
and how cosine similarity can identify semantically
related sentences.

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer("all-MiniLM-L6-v2")

sentences = [
    "I love machine learning.",
    "I enjoy artificial intelligence.",
    "The weather is beautiful today."
]

embeddings = model.encode(sentences)

print("Embedding shape:", embeddings.shape)
```

```python
from sklearn.metrics.pairwise import cosine_similarity

similarity_matrix = cosine_similarity(embeddings)

print(similarity_matrix.round(3))
```

```python
documents = [
    "Machine learning allows computers to learn from data.",
    "Docker packages applications into containers.",
    "Kubernetes orchestrates containerized applications.",
    "Artificial intelligence enables machines to perform intelligent tasks."
]

query = "How do computers learn from examples?"

document_embeddings = model.encode(documents)
query_embedding = model.encode([query])

scores = cosine_similarity(
    query_embedding,
    document_embeddings
)[0]

best_match_index = scores.argmax()

print("Question:", query)
print("Best match:", documents[best_match_index])
print("Similarity:", round(float(scores[best_match_index]), 3))
```

```python
import numpy as np

top_k = 3

top_indices = np.argsort(scores)[::-1][:top_k]

for rank, index in enumerate(top_indices, start=1):
    print(f"{rank}. {documents[index]}")
    print(f"   Similarity: {scores[index]:.3f}")
```

```python
threshold = 0.3
top_k = 3

top_indices = np.argsort(scores)[::-1][:top_k]

for index in top_indices:
    if scores[index] >= threshold:
        print(f"{documents[index]}")
        print(f"Similarity: {scores[index]:.3f}\n")
```
