## 1. Next-Token Generation Loop
**Goal:** token generation step by step step by step.

## Each generated token becomes part of the context for the next prediction.

```python
# Initial state
prompt_tokens = ["Explain", "token", "cost"]
generated_tokens = []

candidate_tokens = ["in", "one", "sentence", "."]
stop_token = "."

# Generation loop
for next_token in candidate_tokens:

    context = prompt_tokens + generated_tokens

    print("Context  ->", " ".join(context))
    print("Selected ->", next_token)

    generated_tokens.append(next_token)

    if next_token == stop_token:
        print("Stop condition met")
        break

# Final result
print("Final text ->", " ".join(prompt_tokens + generated_tokens))
```

**Key Takeaway:**

- LLMs generate text token by token.
- Each generated token extends the context.
- Generation ends when a stop condition is reached.

## 2. Greedy Decoding

**Goal:** how a model selects the next token.

Greedy decoding always selects the token with the highest probability.

```python
# Simulated next-token probabilities

token_probabilities = {
    "Paris": 0.70,
    "London": 0.20,
    "Berlin": 0.10
}

# Greedy decoding

next_token = max(
    token_probabilities,
    key=token_probabilities.get
)

print("Selected token:", next_token)
```

**Key Takeaway:**

Greedy decoding always selects the token with the highest probability.


## 3. Temperature Sampling

**Goal:** show how temperature affects token selection.

Lower temperature makes predictions more predictable.
Higher temperature increases diversity.

```python
import random

token_probabilities = {
    "Paris": 0.70,
    "London": 0.20,
    "Berlin": 0.10
}

tokens = list(token_probabilities.keys())
probabilities = list(token_probabilities.values())

selected_token = random.choices(
    population=tokens,
    weights=probabilities,
    k=1
)[0]

print("Selected token:", selected_token)

temperature = 0.5

adjusted_probabilities = {
    token: probability ** (1 / temperature)
    for token, probability in token_probabilities.items()
}

total = sum(adjusted_probabilities.values())

normalized_probabilities = {
    token: probability / total
    for token, probability in adjusted_probabilities.items()
}

print(normalized_probabilities)

import random

token_probabilities = {
    "Paris": 0.70,
    "London": 0.20,
    "Berlin": 0.10
}

tokens = list(token_probabilities.keys())
probabilities = list(token_probabilities.values())

selected_token = random.choices(
    population=tokens,
    weights=probabilities,
    k=1
)[0]

print("Selected token:", selected_token)

temperature = 0.5

adjusted_probabilities = {
    token: probability ** (1 / temperature)
    for token, probability in token_probabilities.items()
}

total = sum(adjusted_probabilities.values())

normalized_probabilities = {
    token: probability / total
    for token, probability in adjusted_probabilities.items()
}

print(normalized_probabilities)
```

## 4. Top-k and Top-p Sampling

**Goal:** show how to limit candidate tokens.

- Top-k keeps the k most probable tokens.
- Top-p keeps tokens until a cumulative probability threshold is reached.

```python
# Next-token probabilities

token_probabilities = {
    "Paris": 0.70,
    "London": 0.20,
    "Berlin": 0.07,
    "Madrid": 0.02,
    "Rome": 0.01
}

k = 2

# Step 1: Sort by probability (highest first)

sorted_tokens = sorted(
    token_probabilities.items(),
    key=lambda item: item[1],
    reverse=True
)

# Step 2: Keep only the top k candidates

top_k_tokens = sorted_tokens[:k]

print("Top-k candidates:", top_k_tokens)
```

```python
# Top-p sampling

p = 0.95

cumulative_probability = 0
top_p_tokens = []

for token, probability in sorted_tokens:

    # Add probability to cumulative total
    cumulative_probability += probability

    # Save selected token
    top_p_tokens.append(token)

    # Stop when threshold is reached
    if cumulative_probability >= p:
        break

print("Top-p candidates:", top_p_tokens)
```

**Key Takeaway:**

- Top-k selects a fixed number of candidate tokens.
- Top-p selects tokens based on cumulative probability.
- Both methods limit the sampling space.
